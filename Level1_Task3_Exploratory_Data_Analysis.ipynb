Statistical Summary

Data Visualization – Histograms

Histograms were used to examine the distribution of the numerical variables in the dataset and to identify patterns, skewness, and unusual values.

```python
import matplotlib.pyplot as plt

df.hist(figsize=(15, 12), bins=20)

plt.suptitle("Histograms of Dataset Features")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["13"], bins=20, edgecolor="black")

plt.title("Distribution of Column 13")
plt.xlabel("Column 13 Values")
plt.ylabel("Frequency")

plt.show()
```

Scatter Plot

A scatter plot was created to explore the relationship between two numerical variables. This visualization can help identify possible relationships, patterns, clusters, or unusual observations.

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["5"], df["13"])

plt.title("Scatter Plot: Column 5 vs Column 13")
plt.xlabel("Column 5")
plt.ylabel("Column 13")

plt.show()
```

Correlation Analysis

Correlation analysis was performed to examine the strength and direction of relationships between numerical variables in the dataset. Correlation values range from -1 to +1. Values close to +1 indicate a strong positive relationship, values close to -1 indicate a strong negative relationship, and values close to 0 indicate a weak or no linear relationship.

```python
correlation_matrix = df.corr()

correlation_matrix
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 10))

plt.imshow(correlation_matrix, cmap="coolwarm", aspect="auto")
plt.colorbar(label="Correlation")

plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)

plt.title("Correlation Matrix Heatmap")
plt.xlabel("Columns")
plt.ylabel("Columns")

plt.show()
```

```python
correlation_with_column13 = correlation_matrix["13"].sort_values(ascending=False)

print(correlation_with_column13)
```

The correlation matrix was used to identify possible linear relationships between the numerical variables. Some variables showed stronger positive or negative correlations than others. These relationships can help identify features that may be useful for further statistical or predictive analysis. Since the dataset does not provide column descriptions, the correlations were interpreted only at the numerical level without assigning business meanings to the variables.

Linear Regression Analysis
A simple linear regression model was created to examine the relationship between two numerical variables. Column 5 was used as the independent variable, while Column 13 was used as the dependent variable for demonstration purposes.

```python
from sklearn.linear_model import LinearRegression

X = df[["5"]]
y = df["13"]

model = LinearRegression()

model.fit(X, y)
```

```python
y_pred = model.predict(X)

print(y_pred[:10])
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X["5"], y, label="Actual Data")
plt.plot(X["5"], y_pred, label="Regression Line")

plt.title("Linear Regression: Column 5 vs Column 13")
plt.xlabel("Column 5")
plt.ylabel("Column 13")

plt.legend()
plt.show()
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, y_pred)

print("R-squared:", r2)
```

Simple linear regression was applied to explore the relationship between Column 5 and Column 13. The regression line was visualized together with the observed data, and the R-squared value was calculated to evaluate how much of the variation in the dependent variable could be explained by the selected independent variable.
